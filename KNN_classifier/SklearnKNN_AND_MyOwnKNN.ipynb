```python
from scipy.spatial import distance
from collections import Counter
class MyKnn:
  def __init__(self,k=5):
    self.n_neighbors=k
    self.x_train=None
    self.y_train=None

  def fit(self,x_train,y_train):
    self.x_train=x_train
    self.y_train=y_train

  def calculate_distance(self,point1,point2):
    return float(np.linalg.norm(point1-point2))

  def predict(self, x_test):
    y_pred= []

    for i in x_test:
        distances = []

        # calculate distance with every training point
        for j in self.x_train:
            distances.append(self.calculate_distance(i, j))
        n_neighbors=sorted(list(enumerate(distances)),key=lambda x:x[1])[0:self.n_neighbors]
        label=self.majority_count(n_neighbors)
        y_pred.append(label)

    return y_pred

  def majority_count(self,n_neighbors):
    votes=[]
    for i in n_neighbors:
      votes.append(int(self.y_train.iloc[i[0]]))
    votes=Counter(votes)
    return votes.most_common()[0][0]
```

```python
myclassifier=MyKnn(k=5)
```

```python
myclassifier.fit(x_train,y_train)
```

```python
y_pred=myclassifier.predict(x_test)
print(accuracy_score(y_test,y_pred))
```
