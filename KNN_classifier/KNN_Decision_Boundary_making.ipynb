```python
df=pd.read_csv('/content/social_network.csv')
df.head()
```

```python
x=df.loc[:,['Age','EstimatedSalary']]
```

```python
Scalar=StandardScaler()
x=Scalar.fit_transform(x)
```

```python
y=df.iloc[:,-1]
```

```python
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```
