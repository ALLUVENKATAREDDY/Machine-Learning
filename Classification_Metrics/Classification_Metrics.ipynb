```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import shap
```

```python
df = pd.read_csv('/content/heart.csv')

df.head()
```

```python
x_train,x_test,y_train,y_test=train_test_split(df.iloc[:,0:-1],df.iloc[:,-1],test_size=0.2,random_state=2)
```

```python
x_train.head()
```

```python
y_train
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
```

```python
log_classifier=LogisticRegression()
des_classifier=DecisionTreeClassifier()
```

```python
log_classifier.fit(x_train,y_train)
des_classifier.fit(x_train,y_train)
```

```python
y_pred1=log_classifier.predict(x_test)
y_pred2=des_classifier.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
print('Accuracy of logistic regression',accuracy_score(y_test,y_pred1))
print('Accuracy of Decision tree classifier',accuracy_score(y_test,y_pred2))
```

```python
print(confusion_matrix(y_test,y_pred1))
```

```python
from sklearn.metrics import precision_score,recall_score,f1_score
print('For Logistic Regression')
print('----------------------------------------------')
print('Precision:',precision_score(y_test,y_pred1))
print('Recall:',recall_score(y_test,y_pred1))
print('F1 score:',f1_score(y_test,y_pred1))
```

```python
print('For Decision Tree')
print('----------------------------------------------')
print('Precision:',precision_score(y_test,y_pred2))
print('Recall',recall_score(y_test,y_pred2))
print('F1 score:',f1_score(y_test,y_pred2))
```

```python
#By default precision and recall is for positive
#let's try to get the both values for positive and negative using parameter average
print('For Logistic Regression')
print('Precision:',precision_score(y_test,y_pred1,average=None))
print('Recall:',recall_score(y_test,y_pred1,average=None))
```

```python
#By default precision and recall is for positive
#let's try to get the both values for positive and negative using parameter average
print('For Decision Tree')
print('Precision:',precision_score(y_test,y_pred2,average=None))
print('Recall:',recall_score(y_test,y_pred2,average=None))
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred1))
```
